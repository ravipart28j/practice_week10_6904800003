```python
#ข้อ ป.1
def is_valid_score(score):
    return 0 <= score <= 100
def to_grade(score):
    if not is_valid_score(score):
        return None
    if score >= 80:
        return "A"
    elif score >= 70:
        return "B"
    elif score >= 60:
        return "C"
    else:
        return "F"
def print_result(student_id, score):
    if is_valid_score(score):
        grade = to_grade(score)
        print("รหัสนักศึกษา:", student_id)
        print("คะแนน:", score)
        print("เกรด:", grade)
    else:
        print("รหัสนักศึกษา:", student_id)
        print("คะแนนไม่ถูกต้อง กรุณากรอก 0 ถึง 100")
def main():
    print_result("001", 85)
    print_result("002", 55)
    print_result("003", 120)
main()
```

```python
#ข้อ ป.2

def safe_average(total, count):
    if count <= 0:
        return None
    return round(total / count, 2)
def main():
    total = 0
    count = 0
    for i in range(5):
        pm = float(input("กรอกค่า PM2.5 วันที่ " + str(i + 1) + ": "))
        total = total + pm
        count = count + 1
    avg = safe_average(total, count)
    if avg is None:
        print("ไม่มีข้อมูล")
    else:
        print("ค่าเฉลี่ย PM2.5 =", avg)
main()
```
